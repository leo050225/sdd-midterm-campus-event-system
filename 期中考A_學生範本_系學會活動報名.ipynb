````python
# ============================================================
# 第 2 格：建立資料庫與初始資料
# 提醒：每次執行都要 DROP TABLE IF EXISTS 再重建
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
import sqlite3

def init_db(db_name="campus_event.db"):
    conn = sqlite3.connect(db_name)
    cursor = conn.cursor()

    # 建立 activities 資料表
    cursor.execute("""
    CREATE TABLE IF NOT EXISTS activities (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        name TEXT UNIQUE NOT NULL,
        date TEXT NOT NULL,
        capacity INTEGER NOT NULL
    );
    """)

    # 建立 registrations 資料表
    cursor.execute("""
    CREATE TABLE IF NOT EXISTS registrations (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        user_name TEXT NOT NULL,
        activity_id INTEGER NOT NULL,
        people_count INTEGER NOT NULL,
        status TEXT NOT NULL,
        FOREIGN KEY (activity_id) REFERENCES activities(id)
    );
    """)

    # 寫入初始活動資料（若不存在）
    initial_activities = [
        (1, '迎新晚會', '2026-11-15', 50),
        (2, '程式工作坊', '2026-11-20', 30),
        (3, '籃球友誼賽', '2026-11-25', 20)
    ]
    cursor.executemany("""
    INSERT OR IGNORE INTO activities (id, name, date, capacity)
    VALUES (?, ?, ?, ?);
    """, initial_activities)

    conn.commit()
    conn.close()

init_db()
print("資料庫初始化完成！")
````

````python
# ============================================================
# 第 3 格：顯示資料表
# 提醒：定義 show_tables() 並呼叫一次
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
import sqlite3
import pandas as pd

def show_tables(db_name="campus_event.db"):
    conn = sqlite3.connect(db_name)
    print("=== activities 資料表 ===")
    df_act = pd.read_sql_query("SELECT * FROM activities;", conn)
    print(df_act.to_string(index=False))

    print("\n=== registrations 資料表 ===")
    df_reg = pd.read_sql_query("SELECT * FROM registrations;", conn)
    if df_reg.empty:
        print("(目前無報名資料)")
    else:
        print(df_reg.to_string(index=False))
    conn.close()

show_tables()
````

````python
# ============================================================
# 第 4 格：AI 解讀函式
# ============================================================
import os
import time
from groq import Groq

# 讀取金鑰
api_key = None
try:
    from google.colab import userdata
    api_key = userdata.get("GROQ_API_KEY")
except Exception:
    pass

if not api_key:
    api_key = os.environ.get("GROQ_API_KEY")

if not api_key:
    raise ValueError("找不到 GROQ_API_KEY！請在 Colab 鑰匙圖示設定 Secrets，或手動指定 api_key。")

client = Groq(api_key=api_key)

# 清空快取字典
ai_cache = {}

SYSTEM_PROMPT = """你是一個系學會活動報名助理。請依據使用者的自然語言抽取出指令。
必須嚴格遵守以下格式回覆，絕對不能包含任何其他贅詞、標籤、符號或解釋：
輸出格式：動作,活動,人數

規則限制：
- 動作只能是：報名、取消、查詢、統計
- 活動只能是：迎新晚會、程式工作坊、籃球友誼賽、所有、無
- 動作區分重點：
  * 若是詢問「特定活動」的名額或資訊，動作為「查詢」。
  * 若提及「各活動」、「所有活動」、「統計」、「整體人數狀況」，動作為「統計」，且活動輸出「所有」。
- 人數的嚴格規則（請依順序判定）：
  1. 若動作為取消、查詢、統計，且句子「沒有特別提到具體人數數字」，人數必須輸出 0（例如取消整個報名輸出 0；只有明講『一位取消』時才輸出 1）。
  2. 若動作為報名，未提及具體人數時，人數預設輸出 1。
  3. 若句子有明確指定人數數字（如 1、2、3、5 位/人/個），則輸出該正整數。
- 無法判斷或無關問題一律輸出：無法辨識,無,0

範例：
我要報名迎新晚會 2 個人 -> 報名,迎新晚會,2
幫我報名程式工作坊 -> 報名,程式工作坊,1
我不要去迎新晚會了，幫我取消 -> 取消,迎新晚會,0
迎新晚會一位取消 -> 取消,迎新晚會,1
請問程式工作坊還有沒有名額？ -> 查詢,程式工作坊,0
查一下各活動目前的報名人數與統計 -> 統計,所有,0
晚餐要吃什麼？ -> 無法辨識,無,0
"""

def extract_intent(user_input: str, retries: int = 3) -> tuple[str, str, int, str]:
    if user_input in ai_cache:
        raw_res = ai_cache[user_input]
    else:
        raw_res = ""
        for attempt in range(retries):
            try:
                response = client.chat.completions.create(
                    model=MODEL,
                    messages=[
                        {"role": "system", "content": SYSTEM_PROMPT},
                        {"role": "user", "content": user_input}
                    ],
                    reasoning_effort=REASONING_EFFORT,
                    temperature=0.0
                )
                raw_res = response.choices[0].message.content.strip()
                raw_res = raw_res.replace("，", ",").strip()
                ai_cache[user_input] = raw_res
                break
            except Exception as e:
                if "429" in str(e) and attempt < retries - 1:
                    time.sleep(2 ** (attempt + 1))
                else:
                    raw_res = f"ERROR: {str(e)}"
                    break

    # 解析輸出格式：動作,活動,人數
    try:
        parts = [p.strip() for p in raw_res.split(",")]
        if len(parts) == 3:
            action, activity = parts[0], parts[1]
            count = int(parts[2])
            return action, activity, count, raw_res
    except Exception:
        pass

    return "無法辨識", "無", 0, raw_res
````

````python
# ============================================================
# 第 6 格：功能函式
# 提醒：fetchone() 後先檢查 None；用 try/finally 關閉連線；統計用 JOIN＋GROUP BY
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
import sqlite3

def register_activity(user_name: str, activity_name: str, people_count: int, db_name="campus_event.db") -> str:
    # A1: 單次人數檢驗 (1~3 人)
    if people_count < 1 or people_count > 3:
        return "報名失敗：單次報名人數限制為 1～3 人"

    conn = sqlite3.connect(db_name)
    cursor = conn.cursor()

    # 查活動資訊
    cursor.execute("SELECT id, capacity FROM activities WHERE name = ?", (activity_name,))
    act = cursor.fetchone()
    if not act:
        conn.close()
        return f"報名失敗：查無活動「{activity_name}」"
    act_id, capacity = act

    # A2: 每人最多同時報名 2 個活動（已取消不計）
    cursor.execute("""
        SELECT COUNT(DISTINCT activity_id) FROM registrations
        WHERE user_name = ? AND status = 'ACTIVE'
    """, (user_name,))
    current_reg_count = cursor.fetchone()[0]

    # 重複報名同一活動檢查
    cursor.execute("""
        SELECT COUNT(*) FROM registrations
        WHERE user_name = ? AND activity_id = ? AND status = 'ACTIVE'
    """, (user_name, act_id))
    already_registered = cursor.fetchone()[0] > 0

    if already_registered:
        conn.close()
        return "報名失敗：您已報名過此活動"

    if current_reg_count >= 2:
        conn.close()
        return "報名失敗：每人最多同時報名 2 個活動"

    # 計算名額餘額
    cursor.execute("""
        SELECT COALESCE(SUM(people_count), 0) FROM registrations
        WHERE activity_id = ? AND status = 'ACTIVE'
    """, (act_id,))
    occupied = cursor.fetchone()[0]
    remaining = capacity - occupied

    if remaining < people_count:
        conn.close()
        return f"報名失敗：剩餘名額不足（僅剩 {remaining} 位）"

    # 寫入報名紀錄
    cursor.execute("""
        INSERT INTO registrations (user_name, activity_id, people_count, status)
        VALUES (?, ?, ?, 'ACTIVE')
    """, (user_name, act_id, people_count))
    conn.commit()
    conn.close()
    return f"報名成功：{activity_name}，共 {people_count} 人"

def cancel_activity(user_name: str, activity_name: str, db_name="campus_event.db") -> str:
    conn = sqlite3.connect(db_name)
    cursor = conn.cursor()

    cursor.execute("SELECT id FROM activities WHERE name = ?", (activity_name,))
    act = cursor.fetchone()
    if not act:
        conn.close()
        return f"取消失敗：查無活動「{activity_name}」"
    act_id = act[0]

    cursor.execute("""
        SELECT id FROM registrations
        WHERE user_name = ? AND activity_id = ? AND status = 'ACTIVE'
    """, (user_name, act_id))
    records = cursor.fetchall()

    if not records:
        conn.close()
        return f"取消失敗：查無該活動的有效報名記錄"

    cursor.execute("""
        UPDATE registrations SET status = 'CANCELLED'
        WHERE user_name = ? AND activity_id = ? AND status = 'ACTIVE'
    """, (user_name, act_id))
    conn.commit()
    conn.close()
    return f"取消成功：已取消「{activity_name}」的報名"

def query_activity(activity_name: str, db_name="campus_event.db") -> str:
    conn = sqlite3.connect(db_name)
    cursor = conn.cursor()

    cursor.execute("SELECT id, date, capacity FROM activities WHERE name = ?", (activity_name,))
    act = cursor.fetchone()
    if not act:
        conn.close()
        return f"查詢失敗：查無活動「{activity_name}」"
    act_id, act_date, capacity = act

    cursor.execute("""
        SELECT COALESCE(SUM(people_count), 0) FROM registrations
        WHERE activity_id = ? AND status = 'ACTIVE'
    """, (act_id,))
    occupied = cursor.fetchone()[0]
    conn.close()

    remaining = capacity - occupied
    return f"【{activity_name}】日期：{act_date} | 總名額：{capacity} | 目前剩餘名額：{remaining}"

def get_all_statistics(db_name="campus_event.db") -> str:
    # A3: 使用 LEFT JOIN 與 GROUP BY 統計所有活動（包含 0 人報名）
    conn = sqlite3.connect(db_name)
    cursor = conn.cursor()
    query = """
    SELECT
        a.name,
        a.date,
        a.capacity,
        COALESCE(SUM(CASE WHEN r.status = 'ACTIVE' THEN r.people_count ELSE 0 END), 0) AS registered,
        a.capacity - COALESCE(SUM(CASE WHEN r.status = 'ACTIVE' THEN r.people_count ELSE 0 END), 0) AS remaining
    FROM activities a
    LEFT JOIN registrations r ON a.id = r.activity_id
    GROUP BY a.id, a.name, a.date, a.capacity;
    """
    cursor.execute(query)
    rows = cursor.fetchall()
    conn.close()

    res = "=== 全活動報名統計清單 ===\n"
    for name, date, cap, reg, rem in rows:
        res += f"- {name} ({date})：上限 {cap} 人 | 已報名 {reg} 人 | 剩餘 {rem} 人\n"
    return res.strip()
````

````python
# ============================================================
# 第 7 格：主流程與流程測試
# 提醒：流程測試 8～10 步；最後呼叫 show_tables()
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
def process_message(user_name: str, user_input: str) -> str:
    action, activity, count, raw = extract_intent(user_input)

    if action == "無法辨識" or activity == "無":
        return "系統提示：無法辨識您的意圖，請確認活動名稱後重新輸入。"

    if action == "報名":
        return register_activity(user_name, activity, count)
    elif action == "取消":
        return cancel_activity(user_name, activity)
    elif action == "查詢":
        return query_activity(activity)
    elif action == "統計":
        return get_all_statistics()
    else:
        return "系統提示：未知的動作指令。"

# 重置 DB 以確保測試獨立性
import os
if os.path.exists("campus_event.db"):
    os.remove("campus_event.db")
init_db()

test_flow_7_2 = [
    ("小明", "我要報名迎新晚會 2 個人"),
    ("小明", "我要報名程式工作坊 1 個人"),
    ("小明", "我要報名籃球友誼賽 1 個人"),
    ("小明", "幫我取消迎新晚會"),
    ("小明", "我要報名籃球友誼賽 1 個人"),
    ("小華", "我要報名程式工作坊 4 個人"),
    ("小華", "幫我取消籃球友誼賽"),
    ("小華", "請問籃球友誼賽還有名額嗎？"),
    ("小華", "查一下所有活動統計"),
    ("小美", "今天天氣真好")
]

print("=== 執行流程測試 7-2 ===")
for idx, (user, query) in enumerate(test_flow_7_2, 1):
    result = process_message(user, query)
    print(f"[{idx}] {user} 說：'{query}'")
    print(f"    --> 回應：{result}\n")

print("=== 測試完成後的資料表狀態 ===")
show_tables()
````
